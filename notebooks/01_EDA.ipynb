This notebook performs dataset verification and exploratory data analysis for the AI4I 2020 Predictive Maintenance Dataset.

```
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().resolve().parents[1] / "src"))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from config import DATA_PATH, TARGET

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")

df = pd.read_csv(DATA_PATH)
print("Shape:", df.shape)
df.head()
```

```
print("Missing values:")
display(df.isna().sum().to_frame("missing"))

print("\nDuplicates:", df.duplicated().sum())

print("\nData types:")
display(df.dtypes.to_frame("dtype"))
```

```
print("Target distribution:")
display(df[TARGET].value_counts().to_frame("count"))

print("\nTarget percentage:")
display((df[TARGET].value_counts(normalize=True) * 100).round(2).to_frame("percentage"))

plt.figure(figsize=(7,5))
ax = sns.countplot(data=df, x=TARGET)
plt.title("Machine Failure Distribution")
plt.xlabel("Machine Failure")
plt.ylabel("Number of Samples")
for container in ax.containers:
    ax.bar_label(container)
plt.tight_layout()
plt.show()
```

```
numeric_features = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
]

display(df[numeric_features].describe().T)
```

```
for feature in numeric_features:
    plt.figure(figsize=(8,5))
    sns.histplot(data=df, x=feature, hue=TARGET, kde=True, element="step")
    plt.title(f"{feature} Distribution by Failure")
    plt.tight_layout()
    plt.show()
```

```
for feature in numeric_features:
    plt.figure(figsize=(7,5))
    sns.boxplot(data=df, x=TARGET, y=feature)
    plt.title(f"{feature} vs Machine Failure")
    plt.tight_layout()
    plt.show()
```

```
corr = df[numeric_features + [TARGET]].corr()

plt.figure(figsize=(10,7))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix")
plt.tight_layout()
plt.show()
```

```
failure_modes = ["TWF", "HDF", "PWF", "OSF", "RNF"]

display(df[failure_modes].sum().sort_values(ascending=False).to_frame("count"))

plt.figure(figsize=(8,5))
df[failure_modes].sum().sort_values(ascending=False).plot(kind="bar")
plt.title("Failure Mode Counts")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```
